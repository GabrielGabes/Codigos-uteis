```python
import os
from pathlib import Path

def get_folder_size(folder: Path) -> int:
    """Retorna o tamanho total (em bytes) de uma pasta."""
    total_size = 0
    for dirpath, dirnames, filenames in os.walk(folder):
        for f in filenames:
            fp = Path(dirpath) / f
            try:
                total_size += fp.stat().st_size
            except (FileNotFoundError, PermissionError):
                # Ignora arquivos que não podem ser acessados
                pass
    return total_size

def human_readable_size(size_bytes: int) -> str:
    """Converte bytes em uma string legível (KB, MB, GB)."""
    for unit in ['B', 'KB', 'MB', 'GB', 'TB']:
        if size_bytes < 1024:
            return f"{size_bytes:.2f} {unit}"
        size_bytes /= 1024
    return f"{size_bytes:.2f} PB"
```

```python
# Caminho base
base_path = Path(r"C:\Users\gabri\OneDrive\Documentos")

# Lista apenas subpastas imediatas
for folder in base_path.iterdir():
    if folder.is_dir():
        size = get_folder_size(folder)
        print(f"{folder.name}: {human_readable_size(size)}")
```
